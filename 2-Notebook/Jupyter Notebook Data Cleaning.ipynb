```python
df=df.rename(columns={
    'CustomerID':'customer_id',
    'Age':'age',
    'Gender':'gender',
    'Income':'income',
    'CampaignChannel':'campaign_channel',
    'CampaignType':'campaign_type',
    'AdSpend':'ad_spend',
    'ClickThroughRate':'click_through_rate',
    'ConversionRate':'conversion_rate',
    'WebsiteVisits':'website_visits',
    'PagesPerVisit':'pages_per_visit',
    'TimeOnSite':'time_on_site',
    'SocialShares':'social_shares',
    'EmailOpens':'email_opens',
    'EmailClicks':'email_clicks',
    'PreviousPurchases':'previous_purchase',
    'LoyaltyPoints':'loyalty_points',
    'AdvertisingPlatform':'advertising_platform',
    'AdvertisingTool':'advertising_tool',
    'Conversion':'conversion'
})
```

```python
df.columns
```

```python
#creating a columns age_group:
labels=['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
df[['age_group', 'age']]
```

```python
#creating a columns age_group:
labels=['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
df[['age_group', 'age']]
```

```python
df.groupby('age_group')['age'].agg([min, max])
```

```python
#creating a new column income_group:
income_labels=['Low', 'Medium', 'High']
df['income_group']=pd.qcut(df['income'], q=3, labels=income_labels)
df[['income_group', 'income']]
```

```python
df.groupby('income_group')['income'].agg([min, max])
```

```python
df['advertising_platform'].unique()
```

```python
#Both the columns "advertising_platform" and "advertising_tool" contains same value in their each rows.
#I am going to delete them as there's no requirement for those columns for analyzing the data.
df=df.drop('advertising_platform', axis=1)
df=df.drop('advertising_tool', axis=1)
```

```python
df.columns
```

```python
#Renaming the values of the column 'conversion':
df['conversion']=df['conversion'].map({0:'No', 1:'Yes'})
df['conversion']
```

```python
#Exporting Clean Data:
df.to_csv('Digital Marketing Campaign Performance Analysis.csv', index=False)
```
